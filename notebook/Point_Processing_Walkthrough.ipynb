# Point Processing Walkthrough: Gamma Correction and Tone Processing

Narrative notebook covering power law, sRGB transfer, tone curves, and HDR compression.

```
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```
def power_law(img, gamma):
    return np.power(np.clip(img,0,1), gamma)

def srgb_encode(l):
    return np.where(l<=0.0031308, 12.92*l, 1.055*np.power(l,1/2.4)-0.055)

def reinhard(x):
    return x/(1+x)
```

## Worked numerical example
For r=0.25: gamma=0.5 -> 0.5, gamma=2.2 -> approx 0.047.

```
r=0.25
print(r**0.5, r**2.2)
```

```
x=np.linspace(0,1,256,dtype=np.float32)
img=np.tile(x,(128,1))
img3=np.dstack([img,img,img])

out1=power_law(img3,0.6)
out2=srgb_encode(img3)
out3=reinhard(img3*4.0)

fig,ax=plt.subplots(1,4,figsize=(14,3))
for a,t,m in zip(ax,["Input","PowerLaw","sRGB","Reinhard"],[img3,out1,out2,out3]):
    a.imshow(m,cmap="gray",vmin=0,vmax=1); a.set_title(t); a.axis("off")
plt.tight_layout()
```
